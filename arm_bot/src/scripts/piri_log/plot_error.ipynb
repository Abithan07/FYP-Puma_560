```python
import numpy as np
import csv
import os
import argparse
import pandas as pd
import matplotlib
matplotlib.use('TkAgg')   # or Qt5Agg

import matplotlib.pyplot as plt
```

```python
# Function to read a CSV file and return its contents as a DataFrame for column-based access
def read_csv(file_path):
    df = pd.read_csv(file_path)
    df.columns = df.columns.str.strip()
    return df
```

```python
dnn_csv_path = "/data/ros2/ros2_ws2/arm_bot/src/scripts/piri_log/path_004_dnn_log.csv"
# dnn_nofb_csv_path = "/data/ros2/ros2_ws2/arm_bot/src/scripts/piri_log/path_004_dnn_nofb_log.csv"
ctc_csv_path = "/data/ros2/ros2_ws2/arm_bot/src/scripts/piri_log/path_004_ctc_log.csv"
# ctc_nofb_csv_path = "/data/ros2/ros2_ws2/arm_bot/src/scripts/piri_log/path_004_ctc_nofb_log.csv"
pid_csv_path = "/data/ros2/ros2_ws2/arm_bot/src/scripts/piri_log/path_004_pid_log.csv"


dnn_data = read_csv(dnn_csv_path)
ctc_data = read_csv(ctc_csv_path)
pid_data = read_csv(pid_csv_path)

# position error for each controller

dnn_q_des_1 = dnn_data['q_des_1'].astype(float)
dnn_q_act_1 = dnn_data['q_act_1'].astype(float)
dnn_error_1 = np.abs(dnn_q_des_1 - dnn_q_act_1)
ctc_q_des_1 = ctc_data['q_des_1'].astype(float)
ctc_q_act_1 = ctc_data['q_act_1'].astype(float)
ctc_error_1 = np.abs(ctc_q_des_1 - ctc_q_act_1)
pid_q_des_1 = pid_data['q_des_1'].astype(float)
pid_q_act_1 = pid_data['q_act_1'].astype(float)
pid_error_1 = np.abs(pid_q_des_1 - pid_q_act_1)

dnn_q_des_2 = dnn_data['q_des_2'].astype(float)
dnn_q_act_2 = dnn_data['q_act_2'].astype(float)
dnn_error_2 = np.abs(dnn_q_des_2 - dnn_q_act_2)
ctc_q_des_2 = ctc_data['q_des_2'].astype(float)
ctc_q_act_2 = ctc_data['q_act_2'].astype(float)
ctc_error_2 = np.abs(ctc_q_des_2 - ctc_q_act_2)
pid_q_des_2 = pid_data['q_des_2'].astype(float)
pid_q_act_2 = pid_data['q_act_2'].astype(float)
pid_error_2 = np.abs(pid_q_des_2 - pid_q_act_2)

dnn_q_des_3 = dnn_data['q_des_3'].astype(float)
dnn_q_act_3 = dnn_data['q_act_3'].astype(float)
dnn_error_3 = np.abs(dnn_q_des_3 - dnn_q_act_3)
ctc_q_des_3 = ctc_data['q_des_3'].astype(float)
ctc_q_act_3 = ctc_data['q_act_3'].astype(float)
ctc_error_3 = np.abs(ctc_q_des_3 - ctc_q_act_3)
pid_q_des_3 = pid_data['q_des_3'].astype(float)
pid_q_act_3 = pid_data['q_act_3'].astype(float)
pid_error_3 = np.abs(pid_q_des_3 - pid_q_act_3)


# Total Torque of each controller

dnn_tau_total_1 = dnn_data['tau_total_1'].astype(float)
ctc_tau_total_1 = ctc_data['tau_total_1'].astype(float)
pid_tau_total_1 = pid_data['tau_total_1'].astype(float)

dnn_tau_total_2 = dnn_data['tau_total_2'].astype(float)
ctc_tau_total_2 = ctc_data['tau_total_2'].astype(float)
pid_tau_total_2 = pid_data['tau_total_2'].astype(float)

pid_tau_total_3 = pid_data['tau_total_3'].astype(float)
ctc_tau_total_3 = ctc_data['tau_total_3'].astype(float)
dnn_tau_total_3 = dnn_data['tau_total_3'].astype(float)


# Velocity of each controller
dnn_qd_act_1 = dnn_data['qd_act_1'].astype(float)
ctc_qd_act_1 = ctc_data['qd_act_1'].astype(float)
pid_qd_act_1 = pid_data['qd_act_1'].astype(float)

dnn_qd_act_2 = dnn_data['qd_act_2'].astype(float)
ctc_qd_act_2 = ctc_data['qd_act_2'].astype(float)
pid_qd_act_2 = pid_data['qd_act_2'].astype(float)

dnn_qd_act_3 = dnn_data['qd_act_3'].astype(float)
ctc_qd_act_3 = ctc_data['qd_act_3'].astype(float)
pid_qd_act_3 = pid_data['qd_act_3'].astype(float)
```

# Position Error Comparison

```python
# plot the error between the desired and actual positions for each controller
plt.figure(figsize=(12, 8))
plt.subplot(3, 1, 1)
plt.plot(dnn_error_1, label='DeLan + GRU + PID', color='tab:blue')
plt.plot(ctc_error_1, label='Lagrange + PID', color='tab:orange')
plt.plot(pid_error_1, label='PID', color='m')
plt.title('Joint 1 Error')
plt.xlabel('Time Step')
plt.ylabel('Absolute Error (rad)')
plt.legend()
plt.gca().set_facecolor('lightgray')

plt.subplot(3, 1, 2)
plt.plot(dnn_error_2, label='DeLan + GRU + PID', color='tab:blue')
plt.plot(ctc_error_2, label='Lagrange + PID', color='tab:orange')
plt.plot(pid_error_2, label='PID', color='m')
plt.title('Joint 2 Error')
plt.xlabel('Time Step')
plt.ylabel('Absolute Error (rad)')
plt.legend()
plt.gca().set_facecolor('lightgray')

plt.subplot(3, 1, 3)
plt.plot(dnn_error_3, label='DeLan + GRU + PID', color='tab:blue')
plt.plot(ctc_error_3, label='Lagrange + PID', color='tab:orange')
plt.plot(pid_error_3, label='PID', color='m')
plt.title('Joint 3 Error')
plt.xlabel('Time Step')
plt.ylabel('Absolute Error (rad)')
plt.legend()
plt.gca().set_facecolor('lightgray')

# use light gray background for better visibility
plt.tight_layout()
plt.show()
```

# Torque Error Comparison

```python
plt.figure(figsize=(12, 8))
plt.subplot(3, 1, 1)
plt.plot(dnn_tau_total_1, label='DeLan + GRU + PID')
plt.plot(ctc_tau_total_1, label='Lagrange + PID')
plt.plot(pid_tau_total_1, label='PID')
plt.title('Joint 1 Total Torque')
plt.xlabel('Time Step')
plt.ylabel('Total Torque (Nm)')
plt.legend()
plt.gca().set_facecolor('lightgray')

plt.subplot(3, 1, 2)
plt.plot(dnn_tau_total_2, label='DeLan + GRU + PID')
plt.plot(ctc_tau_total_2, label='Lagrange + PID')
plt.plot(pid_tau_total_2, label='PID')
plt.title('Joint 2 Total Torque')
plt.xlabel('Time Step')
plt.ylabel('Total Torque (Nm)')
plt.legend()
plt.gca().set_facecolor('lightgray')

plt.subplot(3, 1, 3)
plt.plot(dnn_tau_total_3, label='DeLan + GRU + PID')
plt.plot(ctc_tau_total_3, label='Lagrange + PID')
plt.plot(pid_tau_total_3, label='PID')
plt.title('Joint 3 Total Torque')
plt.xlabel('Time Step')
plt.ylabel('Total Torque (Nm)')
plt.legend()
plt.gca().set_facecolor('lightgray')

plt.tight_layout()
plt.show()
```

# Calculate Energy

```python
# Power = tau * qd
dnn_power_1 = dnn_tau_total_1 * dnn_qd_act_1
ctc_power_1 = ctc_tau_total_1 * ctc_qd_act_1
pid_power_1 = pid_tau_total_1 * pid_qd_act_1

dnn_power_2 = dnn_tau_total_2 * dnn_qd_act_2
ctc_power_2 = ctc_tau_total_2 * ctc_qd_act_2
pid_power_2 = pid_tau_total_2 * pid_qd_act_2

dnn_power_3 = dnn_tau_total_3 * dnn_qd_act_3
ctc_power_3 = ctc_tau_total_3 * ctc_qd_act_3
pid_power_3 = pid_tau_total_3 * pid_qd_act_3

# Energy = integral of power over time, approximate with cumulative sum
# use absolute value of power to get total energy consumed regardless of direction

dnn_energy_1 = np.cumsum(np.abs(dnn_power_1.values)) * 0.01  # assuming time step of 0.01s
ctc_energy_1 = np.cumsum(np.abs(ctc_power_1.values)) * 0.01
pid_energy_1 = np.cumsum(np.abs(pid_power_1.values)) * 0.01

dnn_energy_2 = np.cumsum(np.abs(dnn_power_2.values)) * 0.01
ctc_energy_2 = np.cumsum(np.abs(ctc_power_2.values)) * 0.01
pid_energy_2 = np.cumsum(np.abs(pid_power_2.values)) * 0.01

dnn_energy_3 = np.cumsum(np.abs(dnn_power_3.values)) * 0.01
ctc_energy_3 = np.cumsum(np.abs(ctc_power_3.values)) * 0.01
pid_energy_3 = np.cumsum(np.abs(pid_power_3.values)) * 0.01

Total_energy_dnn = dnn_energy_1 + dnn_energy_2 + dnn_energy_3
Total_energy_ctc = ctc_energy_1 + ctc_energy_2 + ctc_energy_3
Total_energy_pid = pid_energy_1 + pid_energy_2 + pid_energy_3

print(f"Total Energy DNN: {Total_energy_dnn[-1]:.2f} J")
print(f"Total Energy CTC: {Total_energy_ctc[-1]:.2f} J")
print(f"Total Energy PID: {Total_energy_pid[-1]:.2f} J")
```

```python
# Plot the power for each controller with shaded area under the curve and zero reference line
# convert to numpy arrays for plotting
dnn_p1 = np.asarray(dnn_power_1)
ctc_p1 = np.asarray(ctc_power_1)
pid_p1 = np.asarray(pid_power_1)

dnn_p2 = np.asarray(dnn_power_2)
ctc_p2 = np.asarray(ctc_power_2)
pid_p2 = np.asarray(pid_power_2)

dnn_p3 = np.asarray(dnn_power_3)
ctc_p3 = np.asarray(ctc_power_3)
pid_p3 = np.asarray(pid_power_3)

plt.figure(figsize=(12, 8))

# Joint 1
x_dnn1 = np.arange(len(dnn_p1))
x_ctc1 = np.arange(len(ctc_p1))
x_pid1 = np.arange(len(pid_p1))
plt.subplot(3, 1, 1)
if len(dnn_p1):
    plt.plot(x_dnn1, dnn_p1, label='DeLan + GRU + PID', color='tab:blue')
    plt.fill_between(x_dnn1, dnn_p1, 0, color='tab:blue', alpha=0.2)
if len(ctc_p1):
    plt.plot(x_ctc1, ctc_p1, label='Lagrange + PID', color='tab:orange')
    plt.fill_between(x_ctc1, ctc_p1, 0, color='tab:orange', alpha=0.2)
if len(pid_p1):
    plt.plot(x_pid1, pid_p1, label='PID', color='m')
    plt.fill_between(x_pid1, pid_p1, 0, color='m', alpha=0.2)
plt.axhline(0, color='k', linestyle='--', linewidth=1)
plt.title('Joint 1 Power')
plt.xlabel('Time Step')
plt.ylabel('Power (W)')
plt.legend()
plt.gca().set_facecolor('lightgray')

# Joint 2
x_dnn2 = np.arange(len(dnn_p2))
x_ctc2 = np.arange(len(ctc_p2))
x_pid2 = np.arange(len(pid_p2))
plt.subplot(3, 1, 2)
if len(dnn_p2):
    plt.plot(x_dnn2, dnn_p2, label='DeLan + GRU + PID', color='tab:blue')
    plt.fill_between(x_dnn2, dnn_p2, 0, color='tab:blue', alpha=0.2)
if len(ctc_p2):
    plt.plot(x_ctc2, ctc_p2, label='Lagrange + PID', color='tab:orange')
    plt.fill_between(x_ctc2, ctc_p2, 0, color='tab:orange', alpha=0.2)
if len(pid_p2):
    plt.plot(x_pid2, pid_p2, label='PID', color='m')
    plt.fill_between(x_pid2, pid_p2, 0, color='m', alpha=0.2)
plt.axhline(0, color='k', linestyle='--', linewidth=1)
plt.title('Joint 2 Power')
plt.xlabel('Time Step')
plt.ylabel('Power (W)')
plt.legend()
plt.gca().set_facecolor('lightgray')

# Joint 3
x_dnn3 = np.arange(len(dnn_p3))
x_ctc3 = np.arange(len(ctc_p3))
x_pid3 = np.arange(len(pid_p3))
plt.subplot(3, 1, 3)
if len(dnn_p3):
    plt.plot(x_dnn3, dnn_p3, label='DeLan + GRU + PID', color='tab:blue')
    plt.fill_between(x_dnn3, dnn_p3, 0, color='tab:blue', alpha=0.2)
if len(ctc_p3):
    plt.plot(x_ctc3, ctc_p3, label='Lagrange + PID', color='tab:orange')
    plt.fill_between(x_ctc3, ctc_p3, 0, color='tab:orange', alpha=0.2)
if len(pid_p3):
    plt.plot(x_pid3, pid_p3, label='PID', color='m')
    plt.fill_between(x_pid3, pid_p3, 0, color='m', alpha=0.2)
plt.axhline(0, color='k', linestyle='--', linewidth=1)
plt.title('Joint 3 Power')
plt.xlabel('Time Step')
plt.ylabel('Power (W)')
plt.legend()
plt.gca().set_facecolor('lightgray')

plt.tight_layout()
plt.show()
```
